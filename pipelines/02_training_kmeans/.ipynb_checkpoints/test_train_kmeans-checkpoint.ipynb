```python
import boto3
from io import BytesIO
```

```python
def load_parquet_from_s3(s3_client, bucket, key):
    response = s3_client.get_object(Bucket=bucket, Key=key)
    return pd.read_parquet(BytesIO(response["Body"].read()))
```

```python
# Init client
s3_client = boto3.client("s3")
```

```python
# Load
df = load_parquet_from_s3(
    s3_client,
    bucket="swo-ngoctran-public",
    key="mlops/pipelines/01_preprocessing_kmeans/2024-01/30/23/sample_output1_DWA_processed.parquet"
)
```

```python
from sklearn.cluster import KMeans

# Lấy cột "value" và reshape lại thành 2D array (n_samples, n_features)
X = df["value"].values.reshape(-1, 1)
```

```python
kmeans = KMeans(n_clusters=5, random_state=42)
kmeans.fit(X)
```

```python
df["speed_cluster"] = kmeans.predict(X)
```

```python
print(df[["value", "speed_cluster"]].head())
```

```python
import joblib
import os

os.makedirs("model_dir", exist_ok=True)
joblib.dump(kmeans, "model_dir/model.joblib")
```

```python
import tarfile

with tarfile.open("model.tar.gz", "w:gz") as tar:
    tar.add("model_dir", arcname=".")
```

```python
from sagemaker import get_execution_role
import sagemaker

role = get_execution_role()
sagemaker_session = sagemaker.Session()
bucket = sagemaker_session.default_bucket()

model_artifact = sagemaker_session.upload_data("model.tar.gz", bucket=bucket)
```

```python
from sagemaker.sklearn.model import SKLearnModel
```

```python
sk_model = SKLearnModel(
    model_data=model_artifact,
    role=role,
    entry_point="inference.py",
    framework_version="0.23-1",
    sagemaker_session=sagemaker_session
)
```

```python
from datetime import datetime

endpoint_name = f"kmeans-speed-endpoint-{datetime.now().strftime('%Y%m%d-%H%M%S')}"
```

```python
endpoint_name
```

```python
predictor = sk_model.deploy(
    instance_type="ml.m5.large",
    initial_instance_count=1,
    endpoint_name=endpoint_name
)
```
