```python
import boto3
```

```python

runtime = boto3.client("sagemaker-runtime")

endpoint_name = 'randomcutforest-2025-07-08-01-52-13-778'

payload = "12.3,5.6,77.0,49.8,1.0"    # 5 feature đúng như lúc train
```

```python

response = runtime.invoke_endpoint(
    EndpointName=endpoint_name,
    ContentType="text/csv",
    Body=payload
)

result = response["Body"].read().decode("utf-8")
print("🔍 Prediction result:", result)
```

```python
import json
```

```python
payload = {
    "instances": [
        {"features": [12.3, 5.6, 77.0, 49.8, 1.0]},
        {"features": [15.1, 4.8, 62.3, 52.5, 2.0]}
    ]
}
```

```python

response = runtime.invoke_endpoint(
    EndpointName=endpoint_name,
    ContentType="application/json",
    Accept="application/json",           # trả về JSON
    Body=json.dumps(payload)
)
```

```python
from sagemaker.predictor import Predictor
from sagemaker.serializers import JSONSerializer
from sagemaker.deserializers import JSONDeserializer
import sagemaker

session = sagemaker.Session()
predictor = Predictor(endpoint_name, sagemaker_session=session)
predictor.serializer     = JSONSerializer()
predictor.deserializer   = JSONDeserializer()

scores = predictor.predict(payload)      # payload là dict Python
print(scores)
```
